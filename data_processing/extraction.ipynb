# To check mixed types of columns

```python
df.columns
```

```python
cols = range(len(df.columns))
for idx in cols:
    col_name = df.columns[idx]
    type_counts = df.iloc[:, idx].apply(type).value_counts()
    print(f"Column {idx} ('{col_name}'):")
    print(type_counts, "\n")
```
